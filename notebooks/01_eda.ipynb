```python
import pandas as pd
import numpy as np
import glob
import os

pd.set_option('display.max_columns', None)
```

```python
data_path = "../data/raw/"
all_files = glob.glob(os.path.join(data_path, "*.csv"))

print(f"Found {len(all_files)} files:")
for f in all_files:
    print(f)
```

```python
df_list = []
for filename in all_files:
    temp_df = pd.read_csv(filename)
    temp_df.columns = temp_df.columns.str.strip()
    df_list.append(temp_df)

data = pd.concat(df_list, axis=0, ignore_index=True)
print(data.shape)
```

```python
data.info()
```

```python
print("Missing values per column:\n", data.isnull().sum().sort_values(ascending=False).head(10))
print("\nInfinite values:", np.isinf(data.select_dtypes(include=[np.number])).sum().sum())
```

```python
print(data['Label'].value_counts())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
data['Label'].value_counts().plot(kind='bar')
plt.title('Class Distribution in CICIDS2017')
plt.ylabel('Number of samples')
plt.xticks(rotation=75)
plt.tight_layout()
plt.savefig('../reports/figures/class_distribution.png')
plt.show()
```
